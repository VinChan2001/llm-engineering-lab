```python
conversation = [
    {
        "speaker": "Parent",
        "message": "I want you to choose a stable career like doctor, engineer, or lawyer. Life is hard, and passion alone may not pay the bills."
    },
    {
        "speaker": "Child",
        "message": "I understand you care about my future, but I want to follow my dreams and build a life that feels meaningful to me."
    }
]
```

```python
for turn in conversation:
    print(f'{turn["speaker"]}: {turn["message"]}')
```

```python
def format_conversation(conversation):

    lines = []

    for turn in conversation:

        speaker = turn["speaker"]

        message = turn["message"]

        lines.append(f"{speaker}: {message}")

    return "\n".join(lines)
```

```python
conversation_text = format_conversation(conversation)
print(conversation_text)
```

```python
mediator_system = """
You are the Mediator Agent in a respectful family conversation.

You are speaking with a Parent and a Child about career choices, stability, dreams, and passion.

Your job:
- Understand both sides.
- Do not pick a winner.
- Translate what each side is really worried about.
- Help the conversation stay respectful.
- Suggest a balanced next step.

Your speaking style:
- Calm, warm, and practical.
- Keep replies short: 4 to 6 sentences.
- Speak directly as the mediator.
- Do not use bullet points.
"""
```

```python
conversation_text = format_conversation(conversation)

mediator_user_prompt = f"""
The conversation so far is:

{conversation_text}

Now respond with what you would like to say next, as Mediator.
"""
```

```python
print(mediator_user_prompt)
```

```python
from openai import OpenAI

client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)
```

```python
def call_mediator():
    conversation_text = format_conversation(conversation)

    mediator_user_prompt = f"""
The conversation so far is:

{conversation_text}

Now respond with what you would like to say next, as Mediator.
"""

    messages = [
        {"role": "system", "content": mediator_system},
        {"role": "user", "content": mediator_user_prompt}
    ]

    print("MESSAGES SENT TO MEDIATOR:")
    for message in messages:
        print(message["role"], ":", message["content"])
    print("_" * 80)

    response = client.chat.completions.create(
        model="qwen3:4b",
        messages=messages
    )

    return response.choices[0].message.content
```

```python
mediator_reply = call_mediator()
print(mediator_reply)
```

```python
conversation.append({
    "speaker": "Mediator",
    "message": mediator_reply
})
```

```python
print(format_conversation(conversation))
```

```python
conversation = [
    {
        "speaker": "Alex",
        "message": "I think AI will make traditional college less necessary because people can learn almost anything faster and cheaper now."
    },
    {
        "speaker": "Blake",
        "message": "I disagree a bit. College is not just about information; it also gives structure, credentials, networking, and discipline."
    },
    {
        "speaker": "Charlie",
        "message": "Maybe both are true. AI could change how people learn, but college may still matter for community, trust, and career signaling."
    }
]
```

```python
conversation
```

```python
def format_conversation(conversation):
    lines = []

    for turn in conversation:
        speaker = turn["speaker"]
        message = turn["message"]
        lines.append(f"{speaker}: {message}")

    return "\n".join(lines)
```

```python
conversation_text = format_conversation(conversation)
print(conversation_text)
```

```python
alex_system = """
You are Alex.

You are very optimistic about AI replacing or transforming traditional college education.
You believe people can learn faster, cheaper, and more personally with AI tools.

Your style:
- Confident and energetic.
- Slightly provocative, but not rude.
- Keep your reply short: 3 to 5 sentences.
- Speak directly as Alex.
"""

blake_system = """
You are Blake.

You are skeptical that AI can fully replace traditional college education.
You believe college provides structure, credentials, discipline, mentorship, and networking.

Your style:
- Practical and grounded.
- Respectfully disagree when needed.
- Keep your reply short: 3 to 5 sentences.
- Speak directly as Blake.
"""

charlie_system = """
You are Charlie.

You are the balanced thinker in the conversation.
You try to find the middle ground between Alex and Blake.
You believe AI will change education, but college may still have value in some situations.

Your style:
- Calm and thoughtful.
- Try to connect both sides.
- Keep your reply short: 3 to 5 sentences.
- Speak directly as Charlie.
"""
```

```python
systems = {
    "Alex": alex_system,
    "Blake": blake_system,
    "Charlie": charlie_system
}
```

```python
def build_user_prompt(agent_name, conversation):
    other_agents = [name for name in ["Alex", "Blake", "Charlie"] if name != agent_name]
    other_agents_text = " and ".join(other_agents)

    conversation_text = format_conversation(conversation)

    user_prompt = f"""
You are {agent_name}, in conversation with {other_agents_text}.

The conversation so far is as follows:

{conversation_text}

Now with this, respond with what you would like to say next, as {agent_name}.
"""

    return user_prompt
```

```python
alex_user_prompt = build_user_prompt("Alex", conversation)
print(alex_user_prompt)
```

```python
from openai import OpenAI

client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)
```

```python
def call_agent(agent_name, model="qwen3:4b"):
    system_prompt = systems[agent_name]
    user_prompt = build_user_prompt(agent_name, conversation)

    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ]

    print(f"MESSAGES SENT TO {agent_name.upper()}:")
    for message in messages:
        print(message["role"], ":", message["content"])
    print("_" * 80)

    response = client.chat.completions.create(
        model=model,
        messages=messages
    )

    return response.choices[0].message.content
```

```python
alex_reply = call_agent("Alex")
print(alex_reply)
```

```python
conversation.append({

    "speaker": "Alex",

    "message": alex_reply

})
```

```python
print(format_conversation(conversation))
```

```python
blake_reply = call_agent("Blake")
print(blake_reply)
```

```python
speaking_order = ["Alex", "Blake", "Charlie"]
```

```python
for agent_name in speaking_order:
    reply = call_agent(agent_name)
    print(f"\n{agent_name}:")
    print(reply)

    conversation.append({
        "speaker": agent_name,
        "message": reply
    })
```

```python
num_rounds = 3

for round_num in range(num_rounds):
    print(f"\n========== ROUND {round_num + 1} ==========")

    for agent_name in speaking_order:
        reply = call_agent(agent_name)
        print(f"\n{agent_name}:")
        print(reply)

        conversation.append({
            "speaker": agent_name,
            "message": reply
        })
```
